```python
import requests
import pandas as pd
import numpy as np
from pathlib import Path
from dotenv import load_dotenv 
from tqdm import tqdm
import io
import xlrd
```

```python
PROJECT_ROOT = Path.cwd().parent
BIRTHS_DIR = PROJECT_ROOT/"data"/"raw"/"births"
CASUALTIES_DIR = PROJECT_ROOT/"data"/"raw"/"casualties"
POPULATION_DIR = PROJECT_ROOT/"data"/"raw"/"population"
BIRTHS_DIR.mkdir(parents=True, exist_ok=True)
CASUALTIES_DIR.mkdir(parents=True, exist_ok=True)
POPULATION_DIR.mkdir(parents=True, exist_ok=True)
```

## Source 1: Our World in Data — Sex Ratio at Birth

```python
headers = {"User-Agent": "Mozilla/5.0"}
url = "https://ourworldindata.org/grapher/sex-ratio-at-birth.csv?v=1&csvType=full&useColumnShortNames=true"
response = requests.get(url, headers=headers)

srb_data = pd.read_csv(io.StringIO(response.text))
```

## Source 2: HMD — Births by Sex

```python
all_dfs = []

for file in BIRTHS_DIR.glob("*.txt"):
    df = pd.read_csv(file)
    all_dfs.append(df)

hmd_births = pd.concat(all_dfs, ignore_index=True)
print(hmd_births.shape)
hmd_births.head()
```

## Source 3: HMD — Total Population by Sex

```python
all_dfs = []

for file in POPULATION_DIR.glob("*.txt"):
    df = pd.read_csv(file)
    all_dfs.append(df)

hmd_population = pd.concat(all_dfs, ignore_index=True)
print(hmd_population.shape)
hmd_population.head()
```
